```python
import pandas as pd
import plotly.express as px

# Load the merged dataset
df = pd.read_csv("Analysis_PlanetScope_Sandbox_All - All data.csv")

# Convert date column
df["date"] = pd.to_datetime(df["date"])

# Sort observations
df = df.sort_values("date")

# Create interactive time-series chart
fig = px.line(
    df,
    x="date",
    y="mean",
    color="field_id",
    markers=True,
    title="NDVI Development Across Three Agricultural Fields",
    labels={
        "date": "Observation Date",
        "mean": "Mean NDVI",
        "field_id": "Agricultural Field"
    }
)

fig.update_layout(
    hovermode="x unified",
    template="plotly_white"
)

fig.show()
```

```python
df = df.sort_values(["field_id", "date"])

df["ndvi_change"] = (
    df.groupby("field_id")["mean"].diff()
)

# Find the largest increases and decreases
changes = df.sort_values(
    "ndvi_change",
    key=lambda x: x.abs(),
    ascending=False
)

print(changes[
    ["date", "field_id", "mean", "ndvi_change"]
].head(10))
```

```python
summary = df.groupby("field_id").agg(
    average_ndvi=("mean", "mean"),
    maximum_ndvi=("mean", "max"),
    minimum_ndvi=("mean", "min"),
    observation_count=("mean", "count")
)

print(summary)
```

```python
peak_dates = (
    df.loc[df.groupby("field_id")["mean"].idxmax(),
           ["field_id", "date", "mean"]]
)

print(peak_dates)
```
